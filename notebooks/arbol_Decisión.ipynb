```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.tree import (
    DecisionTreeRegressor,
    plot_tree,
    export_text
)

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from statsmodels.nonparametric.smoothers_lowess import lowess
```

```python
df = pd.read_csv(
    "../data/deals_transformados.csv"
)

df.head()
```

```python
y = df['precio_oferta']

X = df[
    [
        'precio_normal',
        'ahorro_porcentaje',
        'rating_steam',
        'steam_rating_count',
        'metacritic',
        'deal_rating',
        'antiguedad_anios'
    ]
]

```

```python
X = X.fillna(0)
```

```python
import os

os.makedirs(
    "data/graficas",
    exist_ok=True
)
```

```python
splits = {

    "Split_1_80-20": 0.2,

    "Split_2_70-30": 0.3,

    "Split_3_60-40": 0.4
}
```

```python
resultados = []

for nombre_split, test_size in splits.items():

    print(f"\n========== {nombre_split} ==========\n")

    # -----------------------------------
    # TRAIN TEST SPLIT
    # -----------------------------------

    X_train, X_test, y_train, y_test = train_test_split(

        X,
        y,

        test_size=test_size,

        random_state=42
    )

    # -----------------------------------
    # MODELO
    # -----------------------------------

    modelo = DecisionTreeRegressor(

        random_state=42
    )

    modelo.fit(
        X_train,
        y_train
    )

    # -----------------------------------
    # PREDICCIONES
    # -----------------------------------

    y_pred = modelo.predict(X_test)

    # -----------------------------------
    # MÉTRICAS
    # -----------------------------------

    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred
        )
    )

    r2 = r2_score(
        y_test,
        y_pred
    )

    print(f"MAE: {mae:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print(f"R²: {r2:.4f}")

    resultados.append({

        "Split": nombre_split,

        "MAE": mae,

        "RMSE": rmse,

        "R2": r2
    })

    # -----------------------------------
    # RESIDUOS
    # -----------------------------------

    residuos = y_test - y_pred

    # -----------------------------------
    # GRÁFICO 1
    # RESIDUOS VS PREDICHOS
    # -----------------------------------

    plt.figure(figsize=(10,6))

    plt.scatter(
        y_pred,
        residuos,
        alpha=0.6
    )

    lowess_curve = lowess(

        residuos,

        y_pred,

        frac=0.3
    )

    plt.plot(

        lowess_curve[:,0],

        lowess_curve[:,1],

        linewidth=2
    )

    plt.axhline(
        0,
        linestyle='--'
    )

    plt.xlabel("Valores Predichos")

    plt.ylabel("Residuos")

    plt.title(
        f"Residuos vs Predichos\n{nombre_split}"
    )

    plt.savefig(
        f"data/graficas/dt_{nombre_split}.png"
    )

    plt.show()

    # -----------------------------------
    # GRÁFICO 2
    # PREDICCIÓN VS REAL
    # -----------------------------------

    plt.figure(figsize=(8,8))

    plt.scatter(
        y_test,
        y_pred,
        alpha=0.6
    )

    min_val = min(
        y_test.min(),
        y_pred.min()
    )

    max_val = max(
        y_test.max(),
        y_pred.max()
    )

    plt.plot(

        [min_val, max_val],

        [min_val, max_val],

        linestyle='--'
    )

    plt.xlabel("Valores Reales")

    plt.ylabel("Valores Predichos")

    plt.title(
        f"Predicción vs Real\n{nombre_split}"
    )

    plt.show()

    # -----------------------------------
    # VISUALIZACIÓN ÁRBOL
    # -----------------------------------

    fig, ax = plt.subplots(
        figsize=(22,10)
    )

    plot_tree(

        modelo,

        max_depth=4,

        feature_names=X.columns,

        filled=True,

        rounded=True,

        impurity=False,

        precision=2,

        fontsize=8,

        ax=ax
    )

    plt.title(
        f"Árbol Real\n{nombre_split}"
    )

    plt.show()

    # -----------------------------------
    # INFO ÁRBOL
    # -----------------------------------

    print("\nInformación del árbol:")

    print(
        f"Profundidad: "
        f"{modelo.get_depth()}"
    )

    print(
        f"Hojas: "
        f"{modelo.get_n_leaves()}"
    )

    # -----------------------------------
    # ÁRBOL PEDAGÓGICO
    # -----------------------------------

    modelo_viz = DecisionTreeRegressor(

        max_depth=3,

        random_state=42
    )

    modelo_viz.fit(
        X_train,
        y_train
    )

    y_pred_viz = modelo_viz.predict(
        X_test
    )

    r2_viz = r2_score(
        y_test,
        y_pred_viz
    )

    print(
        f"\nR² Árbol pedagógico: "
        f"{r2_viz:.4f}"
    )

    fig, ax = plt.subplots(
        figsize=(24,12)
    )

    plot_tree(

        modelo_viz,

        max_depth=3,

        feature_names=X.columns,

        filled=True,

        rounded=True,

        impurity=False,

        precision=2,

        fontsize=10,

        ax=ax
    )

    plt.title(
        f"Árbol Pedagógico\n{nombre_split}"
    )

    plt.show()

    # -----------------------------------
    # REGLAS
    # -----------------------------------

    reglas = export_text(

        modelo_viz,

        feature_names=list(X.columns)
    )

    print("\nREGLAS DEL ÁRBOL:\n")

    print(reglas)
```

```python
resultados_df = pd.DataFrame(
    resultados
)

resultados_df
```
